```python
# Built-in modules
import json
from pathlib import Path

# Third-party modules
import umap
import numpy as np
import pandas as pd
from tqdm import tqdm
import plotly.express as px
import matplotlib.pyplot as plt
```

```python
def umap2D(X, color: list):
    # UMAP
    reducer = umap.UMAP(n_components=2)
    X_umap = reducer.fit_transform(X)
    # Plot
    df = pd.DataFrame({
        'UMAP1': X_umap[:, 0],
        'UMAP2': X_umap[:, 1], 
        'Color': color,
        'bioIDs': bioIDs
    })
    fig = px.scatter(
        df, 
        x='UMAP1', y='UMAP2', 
        color='Color', 
        title='UMAP 2D',
        hover_name='bioIDs'
        )

    fig.show()
    fig.write_html("interactive_plot.html", include_plotlyjs="cdn")
```

```python
# Load embeddings
file = '/home/asanchez/chonky/ML4MIKC/src/quick/yichun/embeddings.npy'
embeddings = np.load(file)
print(embeddings.shape)
```

```python
# Gather metadata
file = '/home/asanchez/chonky/ML4MIKC/src/quick/yichun/parsed_records.json'
with open(file, 'r') as f:
    records = json.load(f)
bioIDs = [record['bioID'] for record in records]
subtypes = [record['subtype'] for record in records]
```

```python
# Iterate over all JSON files and collect embeddings
X = []
bioIDs = []
subtypes = []
folder = Path('/home/asanchez/ML4MIKC/src/modeling/type1/')
for json_file in folder.glob('*.json'):
    with open(json_file, 'r') as infile:
        data = json.load(infile)
        X.append(data['embedding'])
        bioIDs.append(data['bioID'])
        subtypes.append(data['subtype'])
X = np.array(X)
```

```python
reducer = umap.UMAP(n_components=2)
X_umap = reducer.fit_transform(embeddings)
```

```python
df = pd.DataFrame({
    'UMAP1': X_umap[:, 0],
    'UMAP2': X_umap[:, 1], 
    'Color': subtypes,
    'bioIDs': bioIDs
})
fig = px.scatter(
    df, 
    x='UMAP1', y='UMAP2', 
    color='Color', 
    title='UMAP 2D',
    hover_name='bioIDs'
    )

fig.show()
```
